```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

import joblib
```

```python
X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")

y_train = pd.read_csv("../data/processed/y_train.csv").squeeze()
y_test = pd.read_csv("../data/processed/y_test.csv").squeeze()
```

```python
final_model=LogisticRegression(random_state=42,max_iter=100)
final_model.fit(X_train,y_train)
```

```python
y_pred = final_model.predict(X_test)
y_prob = final_model.predict_proba(X_test)[:,1]
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1-Score :", f1_score(y_test, y_pred))
print("ROC-AUC  :", roc_auc_score(y_test, y_prob))
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
print(classification_report(y_test, y_pred))
```

```python
joblib.dump(
    final_model,
    "../models/logistic_regression_churn_model.pkl"
)
```

```python
loaded_model = joblib.load(
    "../models/logistic_regression_churn_model.pkl"
)
```

```python
loaded_predictions = loaded_model.predict(X_test)
```

```python
np.array_equal(y_pred, loaded_predictions)
```

```python
customer = X_test.iloc[0:1]

print("Prediction:", loaded_model.predict(customer))
print("Probability:", loaded_model.predict_proba(customer))
```
